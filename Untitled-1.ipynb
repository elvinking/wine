```python
from scipy.stats import zscore

z=abs(zscore(wine_sample))

print(z.shape)
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Splitting the dataset into train, validation and test dataset
from sklearn.model_selection import train_test_split

%matplotlib inline


wine_sample.groupby('quality').mean()
```

```python
features = wine_sample.drop('quality', axis=1)
labels = wine_sample['quality']
features
```

```python
def print_results(results):
    print('BEST PARAMS: {}\n'.format(results.best_params_))

    means = results.cv_results_['mean_test_score']
    stds = results.cv_results_['std_test_score']
    for mean, std, params in zip(means, stds, results.cv_results_['params']):
        print('{}(+/-{}) for {}'.format(round(mean, 3), round(std * 2, 3), params))
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(features, labels, test_size=0.2, random_state=42)
x_test, x_val, y_test, y_val = train_test_split(x_test, y_test, test_size=0.5, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import GridSearchCV
import joblib
lr = LinearRegression()
parameters = {
    'fit_intercept':[True, False], 'copy_X':[True, False]
}

cv = GridSearchCV(lr, parameters, cv=5)

cv.fit(x_train, y_train.values.ravel())

print_results(cv)
```

```python
cv.best_estimator_

joblib.dump(cv.best_estimator_, 'LR_model.pkl')
```

```python
#Support Vector Machine Regressor
from sklearn.linear_model import Lasso

ls = Lasso()
parameters = {
    'alpha': [0, 1, 0.01]
}

cv = GridSearchCV(ls, parameters, cv=5)

cv.fit(x_train, y_train.values.ravel())

print_results(cv)
```
